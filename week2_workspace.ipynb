```python
import pandas as pd

# Load the CSV file into a pandas DataFrame
df = pd.read_csv('/content/drive/MyDrive/IA 340/diamonds (2).csv')
```

```python
# Display the first 5 rows of the DataFrame
display(df.head())
```

```python
# Calculate the average price of diamonds per color
average_price_per_color = df.groupby('COLOR')['PRICE'].mean()
# D shows the highest price with color
# Display the result
display(average_price_per_color)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Define the correct order for diamond clarity
clarity_order = ['I1', 'SI2', 'SI1', 'VS2', 'VS1', 'VVS2', 'VVS1', 'IF', 'FL']

# Convert 'CLARITY' column to an ordered categorical type
df['CLARITY'] = pd.Categorical(df['CLARITY'], categories=clarity_order, ordered=True)

# Create the scatter plot
plt.figure(figsize=(12, 7))
sns.scatterplot(data=df, x='WEIGHT', y='PRICE', hue='CLARITY', palette='viridis', s=100, alpha=0.8)

plt.title('Diamond Price vs. Weight by Clarity')
plt.xlabel('Weight (Carats)')
plt.ylabel('Price')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(title='Clarity', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
# Calculate the average price per clarity level
average_price_per_clarity = df.groupby('CLARITY')['PRICE'].mean().reindex(clarity_order)
#The chart demonstrates that the higher clarity of the diamond the higher the final price will be.

plt.figure(figsize=(12, 7))
sns.barplot(x=average_price_per_clarity.index, y=average_price_per_clarity.values, palette='viridis', hue=average_price_per_clarity.index, legend=False)

plt.title('Average Diamond Price by Clarity')
plt.xlabel('Clarity')
plt.ylabel('Average Price')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```
